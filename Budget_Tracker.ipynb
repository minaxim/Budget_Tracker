```python
import pandas as pd

# Load dataset
df = pd.read_csv("Budget_Spending_Data.csv")

# Display records
print("BUDGET SPENDING RECORDS")
print(df.head())

# Total expenditure
total_spending = df["Actual Spending"].sum()
print("\nTotal Spending:\n", total_spending)

# Department-wise expenditure
department_spending = df.groupby("Department")["Actual Spending"].sum()

print("\nDepartment-wise Spending:")
print(department_spending)

# Filter expenses by department
department = input("\nEnter department name: ")

filtered_data = df[
    df["Department"].str.lower() == department.lower()
]

print("\nFiltered Expenses:")
print(filtered_data)

# Export processed data
df.to_csv("budget_spending_report.csv", index=False)

print("\nReport generated successfully!")
```
